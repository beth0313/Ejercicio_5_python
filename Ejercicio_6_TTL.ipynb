## Proceso ETL con dataset público (Iris Dataset)
En el contexto del análisis de datos, una de las tareas más comunes es realizar procesos de ETL ***(Extract, Transform, Load)*** sobre datasets en formato CSV.
Para este ejercicio, el aprendiz deberá trabajar en Google Colab con el dataset Iris, el cual es uno de los datasets más utilizados en ciencia de datos y contiene información de flores con variables numéricas y categóricas.
###El dataset puede ser consumido directamente desde la siguiente URL: ***[Dataset Iris](https://archive.ics.uci.edu/ml/machine-learning-databases/iris/iris.data)***


##Objetivo:
Aplicar un proceso ETL completo en Python usando Pandas:

1. Extraer datos desde una fuente externa (CSV)
2. Transformar los datos (limpieza, tipos, análisis básico)
3. Cargar los datos procesados en un nuevo archivo

##Requerimientos:

##1. Extracción (Extract)
* Cargar el dataset desde la URL usando pandas.
* Asignar nombres de columnas adecuados.



```python
#Cargar el dataset desde la URL usando pandas
import pandas as pd
url = 'https://archive.ics.uci.edu/ml/machine-learning-databases/iris/iris.data'
```

```python
#Asignar nombres de columnas adecuados.
df = pd.read_csv(url, header=None)
df.columns = ['sepal_length', 'sepal_width', 'petal_length', 'petal_width', 'class']
```

##2. Transformación (Transform)
Debe realizar al menos:

1. Verificar estructura (info, head)
2. Renombrar columnas
3. Validar datos nulos
4. Convertir tipos de datos si es necesario
5. Crear una nueva columna (ej: promedio de medidas)
6. Filtrar datos (ej: solo una especie)

```python
#1. Verificar estructura (info, head)
df.info()
```

```python
df.head()
```

```python
#2. Renombrar columnas
df.rename(columns={'sepal_length': 'longitud_sepalo', 'sepal_width': 'ancho_sepalo', 'petal_length': 'longitud_petalo', 'petal_width': 'ancho_petalo', 'class': 'clase'}, inplace=True)
```

```python
#3. Validar datos nulos
df.isnull().sum()
```

```python
#4. Convertir tipos de datos si es necesario
df.dtypes
```

```python
#5. Crea una nueva columna (ej.promedio de medidas)
df['promedio_medidas'] = (df['longitud_sepalo'] + df['ancho_sepalo'] + df['longitud_petalo'] + df['ancho_petalo']) / 4
```

```python
# 6. Filtrar datos (ej: solo una especie)
df_setosa = df[df['clase'] == 'Iris-setosa']
df_setosa.head()
```

##3. Carga (Load)

1. Guardar el dataset limpio en un nuevo archivo CSV
2. Mostrar resumen estadístico

```python
#1. Guardar el dataset limpio en un nuevo archivo CSV
df.to_csv('iris_cleaned.csv', index=False)
```
