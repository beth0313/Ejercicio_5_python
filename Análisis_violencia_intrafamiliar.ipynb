## 4. Revisión de columnas y tipos de datos

## 5. Calidad de los datos

Este apartado revisa dos aspectos importantes de calidad: valores nulos y registros duplicados. Los valores nulos indican informaci?n faltante, mientras que los duplicados pueden repetir casos o registros.


```python
valores_nulos = df.isnull().sum()
print("Nuemero de valores nulos")
print(valores_nulos)
```

## 6. Estadísticas descriptivas

Las estad?sticas descriptivas resumen el comportamiento general de las variables. Para las columnas numericas se revisan medidas como promedio, minimo y maximo. Para las columnas categoricas se revisan conteos y valores frecuentes.


## Estadísticas para Columnas Numéricas:

Aquí revisamos el promedio, mínimo y máximo de las columnas que contienen valores numéricos.

```python
# Seleccionar columnas numéricas
columnas_numericas = df.select_dtypes(include=['int64', 'float64']).columns

print("Estadísticas para columnas numéricas:")
for columna in columnas_numericas:
    print(f"\nColumna: {columna}")
    print(f"  Promedio: {df[columna].mean():.2f}")
    print(f"  Mínimo: {df[columna].min()}")
    print(f"  Máximo: {df[columna].max()}")
```

### Estadísticas para Columnas Categóricas

Para las columnas categóricas, examinamos los conteos de cada valor único y los valores más frecuentes.

```python
# Seleccionar columnas categóricas (tipo 'object')
columnas_categoricas = df.select_dtypes(include=['object']).columns

print("Estadísticas para columnas categóricas:")
for columna in columnas_categoricas:
    print(f"\nColumna: {columna}")
    print(f"  Conteo de valores:")
    display(df[columna].value_counts().head(5)) # Mostrar los 10 más frecuentes
    print(f"  Valor más frecuente: {df[columna].mode()[0]}")
```

## 7. Seleccion de columnas importantes

Este bloque muestra columnas especificas que pueden ser utiles para el analisis, como departamento, genero, grupo etario, armas o medios utilizados y cantidad de casos.


```python
columnas_importantes = ['DEPARTAMENTO', 'GENERO', 'GRUPO ETARIO', 'ARMAS MEDIOS', 'CANTIDAD']
df_importante = df[columnas_importantes]
print("Columnas importantes seleccionadas:")
print(df_importante.head(10))
```

## 8. Valores únicos por variable

Los valores unicos permiten conocer las categor?as disponibles en una columna. Esto ayuda a detectar errores de escritura, categorias repetidas o valores inesperados.


```python
print("Valores únicos por cada columna importante:")
for col in df_importante.columns:
    print(f"\nColumna '{col}':")
    print(df_importante[col].unique())
```

## 9. Frecuencia de categorías

La función `value_counts()` cuenta cu?ntas veces aparece cada categoría dentro de una columna. Esto permite identificar los grupos más frecuentes en el dataset.


```python
print("Frecuencia de categorías por cada columna del dataset original:")

# Seleccionar columnas categóricas (tipo 'object') del DataFrame original 'df'
columnas_categoricas_df = df.select_dtypes(include=['object']).columns

for columna in columnas_categoricas_df:
    print(f"\nColumna '{columna}':")
    # Imprimir todos los conteos de valores para evitar truncar
    print(df[columna].value_counts())
```

## 10. Cálculos sobre la variable cantidad

Esta sección calcula el valor minimo, maximo y promedio de la columna `CANTIDAD`. Estos indicadores ayudan a entender la magnitud de los casos reportados en los registros.


```python
# valor minimo, maximo y promedio de la columna CANTIDAD
print(f"Valor mínimo de CANTIDAD: {df['CANTIDAD'].min()}")
print(f"Valor máximo de CANTIDAD: {df['CANTIDAD'].max()}")
print(f"Promedio de CANTIDAD: {df['CANTIDAD'].mean():.2f}")
```

## 11. Agrupación de datos

En esta sección se agrupan los registros para resumir la información por variables importantes como departamento, género y armas o medios utilizados. La función `groupby()` permite calcular conteos y totales según una o varias categorías.


```python
# Agrupar por DEPARTAMENTO, GENERO, ARMAS MEDIOS y sumar CANTIDAD
agrupacion_variables = df.groupby(['DEPARTAMENTO', 'GENERO', 'ARMAS MEDIOS']).agg(
    Total_Cantidad=('CANTIDAD', 'sum'),
    Conteo_Registros=('CANTIDAD', 'count')
).reset_index()

print("Agrupación por Departamento, Género y Armas/Medios:")
print(agrupacion_variables.head())
```

## 12. Ordenamiento de datos con sort_values

En esta sección se ordenan valores y resultados agrupados usando `sort_values()`. Esta función permite organizar la información de mayor a menor o de menor a mayor para identificar rápidamente los registros o categorías más relevantes.


```python
# ordenan valores y resultados agrupado usando sort_values()
agrupacion_variables_ordenada = agrupacion_variables.sort_values(by='Total_Cantidad', ascending=False)
print("Agrupación ordenada por Total_Cantidad:")
print(agrupacion_variables_ordenada.head())
```

## 13. Operaciones con fechas

En esta sección se convierte la columna `FECHA HECHO` a formato de fecha. Esto permite revisar el rango temporal del dataset y preparar la información para análisis por año, mes o día.


```python
#convierte la columna FECHA HECHO a formato de fecha
df['FECHA HECHO'] = pd.to_datetime(df['FECHA HECHO'], format='%d/%m/%Y')
print("Columna FECHA HECHO convertida a formato de fecha.")
```

```python
# revisar el rango temporal del dataset
fecha_minima = df['FECHA HECHO'].min()
fecha_maxima = df['FECHA HECHO'].max()
print(f"Rango temporal del dataset: Desde {fecha_minima} hasta {fecha_maxima}")
```

## 14. Crear columnas nuevas en el dataset

En esta sección se crean nuevas columnas a partir de la fecha del hecho. Las columnas de año, mes y día facilitan el análisis temporal de los casos reportados.


```python
#crear columnas año, mes o día.
df['Año'] = df['FECHA HECHO'].dt.year
df['Mes'] = df['FECHA HECHO'].dt.month
df['Día'] = df['FECHA HECHO'].dt.day
print("Columnas Año, Mes y Día agregadas al dataset.")
```

## 15. Filtros de información

En esta sección se aplican filtros para seleccionar registros específicos del dataset. Los filtros permiten analizar subconjuntos de información según condiciones como departamento, año, mes, género, grupo etario o municipio.


```python
#aplicar filtros para analizar subconjuntos de informacion según condiciones como departamento, año, mes, género, grupo etario o municipio.

# Ejemplo 1: Filtrar por un departamento específico (e.g., 'BOGOTA')
filtro_departamento = df[df['DEPARTAMENTO'] == 'BOGOTA']
print("\nFiltrado por Departamento 'BOGOTA':")
print(filtro_departamento.head())
```

```python
# Ejemplo 2: Filtrar por un año específico (e.g., 2020)
filtro_año = df[df['Año'] == 2020]
print("\nFiltrado por Año 2020:")
print(filtro_año.head())
```

```python
# Ejemplo 3: Filtrar por un mes específico (e.g., Mes 3 - Marzo)
filtro_mes = df[df['Mes'] == 3]
print("\nFiltrado por Mes 3:")
print(filtro_mes.head())
```

```python

# Ejemplo 4: Combinar múltiples filtros (e.g., 'BOGOTA' en 2021 y 'FEMENINO')
filtro_combinado = df[(df['DEPARTAMENTO'] == 'BOGOTA') & (df['Año'] == 2021) & (df['GENERO'] == 'FEMENINO')]
print("\nFiltrado combinado (BOGOTA, 2021, FEMENINO):")
print(filtro_combinado.head())
```
